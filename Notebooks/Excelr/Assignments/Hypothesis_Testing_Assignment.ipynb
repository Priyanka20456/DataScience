```python
Step:1

# Hypothesis Statement
# The restaurant owners claim that actual weekly operating costs are higher than the theoretical model.

# Null Hypothesis (H₀):
# The mean weekly operating cost follows the model
H0: μ = 1000 + 5X

# Alternative Hypothesis (H₁):
# The mean weekly operating cost is higher than the model
H1: μ > 1000 + 5X

 # This is a right-tailed Z-test

Step:2

 # Calculate the Test Statistic

# Sample mean weekly cost:
x=3050

# Mean NUmber of units produced:
X=600

# Theoretical mean weekly cost:
𝜇=1000+5(600)=4000

# Standard deviation of units:
𝜎𝑋=25

# Cost standard deviation:
σ=5×25=125

# Sample size:
n=25

# Z-test formula:
​Xˉ−μ​/z=σ/n

# Calculation:
𝑧=3050-4000/125/5^2=-950/25=-38

step:3

# Probability and Comparison
# Significance level:
α=0.05

# Critical value for a right-tailed Z-test at α = 0.05:
𝑧-𝑐ritical=1.645

Step:4

# Decision Rule
Reject H₀ if:z>1.645

# Observed test statistic:z=−38
# Since −38 < 1.645, we fail to reject the null hypothesis

Step:5

# Conclusion:
# There is no statistical evidence at the 5% significance level to support the restaurant owners’ claim that weekly operating costs are higher than the theoretical model suggests.
# In fact, the observed mean cost is significantly lower than the expected model cost.
```

```python

import math
from scipy.stats import norm

# Given values
sample_mean = 3050
X_mean = 600
n = 25
alpha = 0.05

# Theoretical mean weekly cost
mu = 1000 + 5 * X_mean

# Standard deviation
sigma_x = 25
sigma = 5 * sigma_x

# Z-test statistic
z = (sample_mean - mu) / (sigma / math.sqrt(n))

# Critical value for right-tailed test
z_critical = norm.ppf(1 - alpha)

# Output results
print("Hypothesis Testing for Weekly Operating Costs\n")
print(f"Theoretical Mean Cost (μ): {mu}")
print(f"Sample Mean Cost (x̄): {sample_mean}")
print(f"Standard Deviation (σ): {sigma}")
print(f"Sample Size (n): {n}\n")

print(f"Z-Test Statistic: {z:.2f}")
print(f"Critical Z-Value (α = 0.05): {z_critical:.3f}\n")

if z > z_critical:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: There is strong evidence that operating costs are higher than the model.")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
    print("Conclusion: There is no evidence that operating costs are higher than the model.")
```
