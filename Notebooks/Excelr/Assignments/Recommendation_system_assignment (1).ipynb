#### Data Preprocessing

```python
# Load the Dataset
import pandas as pd

# Load dataset
df = pd.read_csv("anime.csv")
```

```python
print(df)
```

```python
print("\nColumn Names:")
print(df.columns)
```

```python
# Handle Missing Values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Remove rows with missing values
df = df.dropna()

print("\nDataset shape after handling missing values:", df.shape)
```

```python
# Fill missing categorical values
df['genre'].fillna('Unknown', inplace=True)
df['type'].fillna('Unknown', inplace=True)

# Fill missing numerical values
df['rating'].fillna(df['rating'].mean(), inplace=True)

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
# dataset Exploration
print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

#### Feature Extraction

```python
# Convert Categorical data into numerical form
# One-hot encoding for genre
genre_features = df['genre'].str.get_dummies(sep=', ')

# One-hot encoding for type
type_features = pd.get_dummies(df['type'])

print("\nGenre feature matrix shape:", genre_features.shape)
print("Type feature matrix shape:", type_features.shape)
```

```python
# Normalize numerical features
from sklearn.preprocessing import StandardScaler
import numpy as np

# Convert 'episodes' column to numeric (Unknown → NaN)
df['episodes'] = pd.to_numeric(df['episodes'], errors='coerce')

print("Episodes column converted to numeric.")
print("Missing values after conversion:")
print(df[['episodes', 'rating', 'members']].isnull().sum())

# Fill missing values with column mean
df['episodes'].fillna(df['episodes'].mean(), inplace=True)
df['rating'].fillna(df['rating'].mean(), inplace=True)
df['members'].fillna(df['members'].mean(), inplace=True)

print("\nMissing values handled successfully!")

# Select numerical features
numerical_features = df[['episodes', 'rating', 'members']]

# Normalize numerical features
scaler = StandardScaler()
numerical_scaled = scaler.fit_transform(numerical_features)

print("\nNumerical features normalized successfully!")
print("Scaled feature sample:")
print(numerical_scaled[:5])
```

```python
# Final Feature Matrix
import numpy as np

feature_matrix = np.hstack((
    genre_features.values,
    type_features.values,
    numerical_scaled
))

print("\nFinal Feature Matrix Shape:", feature_matrix.shape)
```

#### Recommendation system using cosine similarity

```python
# Compute Cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(feature_matrix)

print("\nCosine Similarity Matrix created")
print("Cosine Similarity Matrix Shape:", cosine_sim.shape)
```

```python
# Recommendation Function
def recommend_anime(anime_name, top_n=5, threshold=0.5):
    if anime_name not in df['name'].values:
        print("Anime not found in dataset")
        return

    anime_index = df[df['name'] == anime_name].index[0]

    similarity_scores = list(enumerate(cosine_sim[anime_index]))
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)

    filtered_scores = [
        score for score in similarity_scores
        if score[1] >= threshold and score[0] != anime_index
    ]

    print(f"\nTop {top_n} anime recommendations for '{anime_name}':\n")

    for i, score in filtered_scores[:top_n]:
        print(f"{df.iloc[i]['name']}  --> Similarity Score: {score:.2f}")
```

```python
# Example recommendation
recommend_anime("Naruto", top_n=5, threshold=0.6)
```
