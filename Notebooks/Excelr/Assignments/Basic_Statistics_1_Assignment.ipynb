```python
import pandas as pd

df = pd.read_csv("sales_data_with_discounts.csv")
df.head()
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64'])
numerical_cols.columns
```

```python
stats = pd.DataFrame({
    'Mean': numerical_cols.mean(),
    'Median': numerical_cols.median(),
    'Mode': numerical_cols.mode().iloc[0],
    'Standard Deviation': numerical_cols.std()
})

stats
```

```python
import matplotlib.pyplot as plt

numerical_cols.hist(figsize=(10, 8))
plt.tight_layout()
plt.show()
```

```python
for col in numerical_cols.columns:
    plt.figure()
    plt.boxplot(df[col])
    plt.title(f'Box Plot of {col}')
    plt.show()
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category'])
categorical_cols.columns
```

```python
for col in categorical_cols.columns:
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Bar Chart of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()
```

#### Conclusion

```python
# Descriptive statistics revealed central tendency and variability in sales-related variables.

# Histograms highlighted skewness and potential outliers in numerical columns.

# Box plots confirmed the presence of extreme values, particularly in sales and profit.

# Bar charts showed imbalanced distributions among categorical variables.

# These insights are crucial for data preprocessing, such as outlier treatment, normalization, and feature engineering before advanced analytics or modeling.
```
