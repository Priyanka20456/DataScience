#### Handling Missing Values

```python
# Import required libraries
import pandas as pd
import numpy as np

#  Load the dataset
file_path = "adult_with_headers (1).csv"
df = pd.read_csv(file_path)

# Display initial shape
print("Initial dataset shape:", df.shape)

#  Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

#  Check missing values BEFORE imputation
print("\nMissing values before imputation:")
print(df.isnull().sum())

# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

print("\nNumerical Columns:")
print(list(num_cols))

print("\nCategorical Columns:")
print(list(cat_cols))

#  Impute numerical columns with MEDIAN
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

#  Impute categorical columns with MODE
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

#  Verify missing values AFTER imputation
print("\nMissing values after imputation:")
print(df.isnull().sum())

#  Final dataset info
print("\nFinal dataset info:")
print(df.info())
```

```python
# Scaling Numerical Features
# Standard Scaling
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()
df_std_scaled = df.copy()
df_std_scaled[num_cols] = std_scaler.fit_transform(df[num_cols])

#  Display results
print("Standard Scaled Numerical Features:")
print(df_std_scaled[num_cols].head())

#  Verify mean ≈ 0 and std ≈ 1
print("\nMean of scaled features:")
print(df_std_scaled[num_cols].mean())

print("\nStandard deviation of scaled features:")
print(df_std_scaled[num_cols].std())
```

```python
# Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler

mm_scaler = MinMaxScaler()
df_mm_scaled = df.copy()
df_mm_scaled[num_cols] = mm_scaler.fit_transform(df[num_cols])

 # Apply Min-Max Scaling
mm_scaler = MinMaxScaler()
df_mm_scaled = df.copy()
df_mm_scaled[num_cols] = mm_scaler.fit_transform(df[num_cols])

# Display scaled numerical features
print("Min-Max Scaled Numerical Features (0–1 range):")
print(df_mm_scaled[num_cols].head())

# Verify min and max values
print("\nMinimum values after scaling:")
print(df_mm_scaled[num_cols].min())

print("\nMaximum values after scaling:")
print(df_mm_scaled[num_cols].max())







```

```python
# Comparison
# | Scaling Method | When to Use                            |
# | StandardScaler | Normal distribution, no fixed bounds   |
# | MinMaxScaler   | Neural networks, distance-based models |
```

#### Encoding Techniques

```python
# Create cardinality table for categorical variables
cardinality_df = pd.DataFrame({
    'Column Name': cat_cols,
    'Unique Categories': [df[col].nunique() for col in cat_cols]
}).sort_values(by='Unique Categories')

# Display cardinality table
print("\nCategorical Variable Cardinality:")
print(cardinality_df)

#  Separate low and high cardinality variables
low_cardinality = cardinality_df[
    cardinality_df['Unique Categories'] < 5
]['Column Name'].tolist()

high_cardinality = cardinality_df[
    cardinality_df['Unique Categories'] >= 5
]['Column Name'].tolist()

#  Display results clearly
print("\nLow Cardinality Categorical Variables (<5 categories):")
print(low_cardinality)
print("\nHigh Cardinality Categorical Variables (≥5 categories):")
print(high_cardinality)



```

```python
# One Hot Encoding
# Import libraries
import pandas as pd
import numpy as np
cardinality_df = pd.DataFrame({
    'Column': cat_cols,
    'Unique_Values': [df[col].nunique() for col in cat_cols]
})

low_cardinality = cardinality_df[
    cardinality_df['Unique_Values'] < 5
]['Column'].tolist()

print("Low-cardinality categorical columns (<5 categories):")
print(low_cardinality)

# Apply One-Hot Encoding (ONLY ONCE)
df_ohe = pd.get_dummies(df, columns=low_cardinality, drop_first=True)

# Show output clearly
print("\nShape before One-Hot Encoding:", df.shape)
print("Shape after One-Hot Encoding:", df_ohe.shape)

new_columns = list(set(df_ohe.columns) - set(df.columns))
print("\nNew columns created by One-Hot Encoding:")
print(new_columns)

print("\nSample rows of encoded dataset:")
print(df_ohe.head())
```

```python
# Label Encoding (Categories ≥ 5)
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
for col in high_cardinality:
    df[col] = le.fit_transform(df[col])

# Apply Label Encoding
le = LabelEncoder()

for col in high_cardinality:
    df[col] = le.fit_transform(df[col])

print("\nData types after Label Encoding:")
print(df[high_cardinality].dtypes)

print("\nSample rows after Label Encoding:")
print(df[high_cardinality].head())
```

```python
# | Encoding | Pros                  | Cons                |
# | One-Hot  | No ordinal assumption | High dimensionality |
# | Label    | Memory efficient      | Implies false order |
```

#### Feature Engineering

```python
# Work experience Intensity
# Find the correct hours column
hours_col = [col for col in df.columns if 'hour' in col.lower()][0]
print("Using column:", hours_col)

# Create new feature
df['hours_per_year'] = df[hours_col] * 52

# Display output
print(df[['hours_per_year']].head())

# Rationale: Annual work effort better represents earning capacity.
```

```python
# Educating Level Bucketing

print(df.columns)

```

```python
import pandas as pd
import numpy as np

# Handle missing values
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# PRINT ALL COLUMN NAMES (DEBUG – IMPORTANT)
print("\nAll column names in dataset:")
print(list(df.columns))

#  AUTO-DETECT education numeric column
edu_candidates = [
    col for col in df.columns
    if 'education' in col.lower() and df[col].dtype != 'object'
]

print("\nDetected education numeric column(s):", edu_candidates)

# Select the first detected column
edu_col = edu_candidates[0]

#  Create education level feature
df['education_level'] = df[edu_col].apply(
    lambda x: 'Low' if x <= 8 else 'Medium' if x <= 12 else 'High'
)

#  One-Hot Encode education_level
df = pd.get_dummies(df, columns=['education_level'], drop_first=True)
print("\nEducation level dummy variables (sample):")
print(df[['education_level_Medium']].head())

# Rationale: Groups education into meaningful income-based segments.
```

```python
# Transformation Of Skewed Features
df[num_cols].skew().sort_values(ascending=False)
```

```python
# Find the correct capital gain column
cap_gain_col = [col for col in df.columns if 'capital' in col.lower() and 'gain' in col.lower()][0]

print("Using capital gain column:", cap_gain_col)

# Apply log transformation
df['capital_gain_log'] = np.log1p(df[cap_gain_col])
print("\nOriginal vs Log-transformed capital gain:")
print(df[[cap_gain_col, 'capital_gain_log']].head())
```
