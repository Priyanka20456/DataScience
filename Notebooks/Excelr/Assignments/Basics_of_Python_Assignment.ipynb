```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print(f"What is the product of {num1} and {num2}?")
user_answer = int(input("Enter your answer: "))

correct_answer = num1 * num2

if user_answer == correct_answer:
    print("Correct! Well done ")
else:
    print(f"Incorrect. The correct answer is {correct_answer}.")
```

```python

room_charge_per_day = 2000
doctor_fee = 1500
lab_test_cost = 300

days_stayed = 4

number_of_tests = int(input("Enter number of lab tests taken: "))
medicine_charges = float(input("Enter total medicine charges (₹): "))

room_charges = days_stayed * room_charge_per_day
lab_charges = number_of_tests * lab_test_cost

total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

print("\n--- Hospital Bill Details ---")
print(f"Room Charges: ₹{room_charges}")
print(f"Doctor Fee: ₹{doctor_fee}")
print(f"Lab Charges: ₹{lab_charges}")
print(f"Medicine Charges: ₹{medicine_charges}")
print(f"Total Bill (Before Discount): ₹{total_bill}")
print(f"Discount Applied: ₹{discount}")
print(f"Final Bill Amount: ₹{final_bill}")
```
