```python
# Create a list of 5 integers
nums = [5, 10, 15, 20, 25]
print("Original list:", nums)

# append
nums.append(30)
print("After append:", nums)

# extend
nums.extend([35, 40])
print("After extend:", nums)

# insert
nums.insert(2, 12)
print("After insert:", nums)

# remove
nums.remove(20)
print("After remove:", nums)

# pop
nums.pop()
print("After pop:", nums)

# index
print("Index of 15:", nums.index(15))

# count
print("Count of 10:", nums.count(10))

# sort
nums.sort()
print("After sort:", nums)

# reverse
nums.reverse()
print("After reverse:", nums)

# clear
nums.clear()
print("After clear:", nums)
```

```python
import pandas as pd
import numpy as np

# Create a small DataFrame manually
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jane'],
    'Age': [25, 30, 22, 28, 35, 40, 30, 28, 22, 35],
    'City': ['NY', 'LA', 'NY', 'Chicago', 'LA', 'NY', 'Chicago', 'LA', 'NY', 'LA'],
    'Salary': [50000, 60000, np.nan, 55000, 70000, 80000, np.nan, 65000, 50000, 70000]
}

df = pd.DataFrame(data)

# Display DataFrame info
print("Info:")
print(df.info(), "\n")

# Describe the DataFrame (numerical columns)
print("Describe:")
print(df.describe(), "\n")

# Select 'Name' and 'City' columns
print("Name and City columns:")
print(df[['Name', 'City']], "\n")

# Drop 'City' column
df_dropped = df.drop('City', axis=1)
print("After dropping 'City':")
print(df_dropped, "\n")

# Fill missing values in 'Salary' with mean
salary_mean = df['Salary'].mean()
df['Salary'].fillna(salary_mean, inplace=True)
print("After filling missing Salary with mean:")
print(df, "\n")

# Remove duplicate rows
df_no_duplicates = df.drop_duplicates()
print("After removing duplicates:")
print(df_no_duplicates)
```
